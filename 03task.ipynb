## 📋 Формирование блока ответов

Запустите ячейку ниже, чтобы сформировать блок с вашими ответами. Скопируйте полученный текст и отправьте его преподавателю.

```python
print("=== БЛОК ОТВЕТОВ ===")
print()
for i in range(1, 16):
    var_name = f'result_{i}'
    if var_name in globals():
        print(f"- Задание {i} - {globals()[var_name]}")
    else:
        print(f"- Задание {i} - НЕ ВЫПОЛНЕНО")
print()
print("=== КОНЕЦ БЛОКА ОТВЕТОВ ===")
```
